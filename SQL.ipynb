```python
import pandas as pd
import sqlite3
df=pd.DataFrame({
    "ner":["Бат","Сараа","Болд","Дулмаа","Ганаа"],
    "nas":[25,30,22,28,35],
    "hot":["УБ","Дархан","УБ","Эрдэнэт","УБ"],
    "tsalin":[1500000, 2000000, 1200000, 1800000, 2500000]
})
conn=sqlite3.connect(":memory:")
df.to_sql("ajilchid", conn, index=False, if_exists="replace")
print("Бүгд:")
print(pd.read_sql("SELECT*FROM ajilchid",conn))
print("\nНэр, цалин:")
print(pd.read_sql("SELECT ner, tsalin FROM ajilchid",conn))
print("\nӨндөр цалинтай:")
print(pd.read_sql("SELECT*FROM ajilchid WHERE tsalin>1500000", conn))
print("\nЦалингаар:")
print(pd.read_sql("SELECT * FROM ajilchid ORDER BY tsalin DESC", conn))
print("\nНэр, хот:")
print(pd.read_sql("SELECT ner, hot FROM ajilchid",conn))
print(pd.read_sql("SELECT*FROM ajilchid WHERE nas>28",conn))
print(pd.read_sql("SELECT*FROM ajilchid WHERE hot='УБ'", conn))
print(pd.read_sql("SELECT*FROM ajilchid ORDER BY nas DESC",conn))
print(pd.read_sql("SELECT*FROM ajilchid WHERE tsalin>1500000 AND hot='УБ'",conn))
```
